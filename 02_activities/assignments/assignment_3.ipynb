#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

> Your answer here ... 
> Standardizing features is critically important before applying clustering algorithms such as K-Means because these algorithms rely directly on distance calculations, typically Euclidean distance, to determine similarity between data points. When features are measured on different scales, those with larger numeric ranges dominate the distance computation and therefore disproportionately influence the clustering results.

In the Wine dataset, for example, some features such as proline range in the hundreds or thousands, while others such as hue or flavanoids are typically between 0 and 5. If the data are not standardized, K-Means will mostly cluster wines based on proline and similar large-scale variables, while largely ignoring chemically meaningful features with smaller ranges. This leads to clusters that reflect measurement scale rather than true underlying structure in the data.

Standardization (e.g., transforming each feature to have mean 0 and standard deviation 1) places all variables on the same scale. This ensures that each feature contributes equally to the distance calculations, allowing K-Means to detect patterns based on the actual relationships between chemical properties rather than their units of measurement.

Using unstandardized data can result in misleading clusters, poor separation between wine classes, and unstable centroids. It can also cause important but small-scale features to be ignored entirely, reducing both the accuracy and interpretability of the clustering. Therefore, standardization is essential for obtaining meaningful and reliable clustering results when using distance-based algorithms like K-Means.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...

X = wine_df.values  # shape: (n_samples, n_features)

# Fit K-Means with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X)

# Assign cluster labels back to the original DataFrame
clustered_wine_data = wine_df.copy()
clustered_wine_data["Cluster"] = cluster_labels

print(clustered_wine_data.head())

# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

> Your answer here...
> One common method is the Elbow Method. It runs K-Means for a range of k values (e.g., k=1 to 10) and plots the within-cluster sum of squares (WCSS) or inertia against k. As k increases, inertia always decreases, but after a certain point the improvement becomes marginal. The “elbow” point—where the rate of improvement sharply slows—provides a practical estimate of the optimal number of clusters because it balances cluster compactness (lower inertia) with model simplicity (avoiding unnecessary clusters).

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...

# Extract the column as a NumPy array
color_intensity = wine_df["color_intensity"].values

# Point estimate from the original sample
mean_color_intensity = np.mean(color_intensity)

# Bootstrapping parameters
np.random.seed(123)
n_bootstrap = 10000
n = len(color_intensity)

# Generate bootstrap sample means
bootstrap_means = np.empty(n_bootstrap)
for i in range(n_bootstrap):
    sample = np.random.choice(color_intensity, size=n, replace=True)
    bootstrap_means[i] = np.mean(sample)

# 90% confidence interval using percentile method (5th and 95th percentiles)
lower_bound = np.percentile(bootstrap_means, 5)
upper_bound = np.percentile(bootstrap_means, 95)

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

>  Your answer...
> We use bootstrapping to approximate the sampling distribution of the mean when we only have one observed dataset and may not want to rely on strong parametric assumptions (e.g., strict normality). By repeatedly resampling the observed color_intensity values with replacement and recomputing the mean, bootstrapping shows how the mean would vary across many hypothetical samples from the same underlying population. This helps us understand the uncertainty and stability of the sample mean as an estimate of the true population mean.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

> Your answer...
> The bootstrap confidence interval provides a plausible range of values for the true population mean based on the observed data. It directly summarizes the variability in the bootstrap means:

> * A narrow interval indicates low variability and suggests the mean estimate is relatively precise.

> * A wide interval indicates higher variability and suggests greater uncertainty in the mean estimate.

In other words, the CI turns the bootstrap distribution into an interpretable statement about precision and estimation uncertainty, rather than relying on a single point estimate.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

> Your answer here...
> The bootstrap distribution of the mean typically appears as a tight, roughly bell-shaped cloud centered near the original sample mean. If the bootstrap means show a relatively small spread and the 90% confidence interval is reasonably narrow, it indicates that the estimated mean color intensity is stable—small changes in the sample (via resampling) do not cause large swings in the mean. This supports the conclusion that the sample mean is a reliable summary of central tendency for color_intensity in this dataset. If the CI were wide or the bootstrap means were highly dispersed, that would suggest the mean is more sensitive to sampling variability and therefore less precise.
